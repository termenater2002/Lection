```python
import random
from enum import StrEnum
from typing import Self
from rich import print as rprint

class CharacterTypes(StrEnum):
    GUARDIAN = "Omori"
    EXPLORER = "Mari"
    ARCHER = "Basil"
    WARRIOR = "Aubrey"
    HEALER = "Hero"

class Character():
    def __init__(self, name: str, hp: int, maxhp: int, team_color: str = "white"):
        self.name = name
        self.hp = hp
        self.maxhp = maxhp
        self.team_color = team_color
        
        character_roles = {
            "Omori_Guardian": CharacterTypes.GUARDIAN, "Omori_Defender": CharacterTypes.GUARDIAN,
            "Mari_Scout": CharacterTypes.EXPLORER, "Mari_Runner": CharacterTypes.EXPLORER,
            "Aubrey_Warrior": CharacterTypes.WARRIOR, "Aubrey_Brawler": CharacterTypes.WARRIOR,
            "Basil_Shooter": CharacterTypes.ARCHER, "Basil_Marksman": CharacterTypes.ARCHER,
            "Hero_Healer": CharacterTypes.HEALER, "Hero_Buffer": CharacterTypes.HEALER,
        }
        self.character_type = character_roles[name]
        
    def colored_str(self):
        status = "TOAST" if self.hp <= 0 else f"{self.hp} HP"
        return f"[{self.team_color}]{self.name}[/] ({self.character_type}) - {status}"

    def attack(self, other: "Character") -> bool:
        damage_ranges = {
            CharacterTypes.GUARDIAN: (30, 50),
            CharacterTypes.EXPLORER: (20, 35),
            CharacterTypes.WARRIOR: (25, 45),
            CharacterTypes.ARCHER: (35, 60),
            CharacterTypes.HEALER: (20, 30),
        }
        
        min_damage, max_damage = damage_ranges[self.character_type]
        damage = random.randint(min_damage, max_damage)
        print(f"{self.name} attacks {other.name} na {damage}")
        other.hp -= damage
        if other.hp <= 0:
            other.hp = 0
            self.hp = min(self.hp + 10, self.maxhp)
            print(f"{other.name} umer ot {self.name}!")
            return True
        return False

class Team():
    def __init__(self, name: str, size: int = 5):
        self.name = name
        self.characters = []
        character_names = [
            "Omori_Guardian", "Omori_Defender", "Mari_Scout", "Mari_Runner", 
            "Aubrey_Warrior", "Aubrey_Brawler", "Basil_Shooter", "Basil_Marksman", 
            "Hero_Healer", "Hero_Buffer"
        ]
        color = "black" if name == "Omori" else "purple" if name == "Sunny" else "purple"
        
        for _ in range(size):
            character_name = random.choice(character_names)
            character_hp = random.randint(100, 150)
            character = Character(name=character_name, hp=character_hp, maxhp=character_hp, team_color=color)
            self.characters.append(character)
            
    def print_squad(self):
        rprint(f"Squad {self.name}:")
        for character in self.characters:
            rprint(f"  - {character.colored_str()}")
            
    def get_alive_characters(self):
        return [character for character in self.characters if character.hp > 0]
    
    def is_defeated(self):
        return len(self.get_alive_characters()) == 0
    
    def attack(self, other_squad: "Team"):
        attackers = self.get_alive_characters()
        defenders = other_squad.get_alive_characters()
        if not attackers or not defenders:
            return
        attacker = random.choice(attackers)
        defender = random.choice(defenders)
        attacker.attack(defender)
        
squad1 = Team("Omori")
squad2 = Team("Sunny")

rprint("     teams")
squad1.print_squad()
squad2.print_squad()

round_counter = 1
while not squad1.is_defeated() and not squad2.is_defeated():
    rprint(f"      Round {round_counter}")
    squad1.attack(squad2)
    if squad2.is_defeated():
        break
    squad2.attack(squad1)
    round_counter += 1
    
rprint("      It's over")
if squad1.is_defeated():
    rprint("Sunny Team wins")    
else:
    rprint("Omori Team wins")
rprint("Status:")
squad1.print_squad()  
squad2.print_squad()
```
